```python
with open(r"C:\Users\Admin\Downloads\archive (4)\next_word_predictor.txt") as f:
  data = f.read()
```

```python
data
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([data])
```

```python
tokenizer.word_index
```

```python
input_seq = []

for paragrahs in data.split('\n\n'):
    for sentence in paragrahs.split('. '):
        tokenized_output = tokenizer.texts_to_sequences([sentence])[0]
        print(tokenized_output)
    for i in range(1,len(tokenized_output)):
        input_seq.append(tokenized_output[:i+1])
```

```python
max_lenght = max([len(i) for i in input_seq])
```

```python
from keras.preprocessing.sequence import pad_sequences
```

```python
padded_data = pad_sequences(input_seq,maxlen=max_lenght,padding='pre')
```

```python
X = padded_data[:,:-1]
y = padded_data[:,-1]
```

```python
from keras.utils import to_categorical
```

```python
y = to_categorical(y, num_classes=max(y)+1)
```

```python
from keras.models import Sequential
from keras.layers import Input,Embedding,LSTM,Dense,Dropout,Bidirectional
```

```python
num_classes = y.shape[1]

model = Sequential()
model.add(Input(shape=(22,)))

model.add(Embedding(input_dim=num_classes, output_dim=100))
# input_dim must match vocabulary size (or max token index + 1)

model.add(Bidirectional(LSTM(150)))
model.add(Dropout(0.2))

model.add(Dense(num_classes, activation='softmax'))  # match labels

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',
             loss='categorical_crossentropy',
             metrics = ['accuracy'])
```

```python
model.fit(X,y,epochs=100)
```

```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 🔥 Temperature Sampling Function
def sample_with_temperature(preds, temperature=0.8):
    preds = np.asarray(preds).astype('float64')

    preds = np.log(preds + 1e-8) / temperature   # avoid log(0)
    exp_preds = np.exp(preds)
    preds = exp_preds / np.sum(exp_preds)

    return np.random.choice(len(preds), p=preds)


# 🔥 Top-K Sampling Function
def top_k_sampling(preds, k=5):
    indices = np.argsort(preds)[-k:]
    probs = preds[indices]
    probs = probs / np.sum(probs)

    return np.random.choice(indices, p=probs)


# 🔥 Final Prediction Function
def predict_next_word(seed_text, next_words=3, temperature=0.8, k=5):

    ignore_words = {"the", "a", "an"}   # optional filter

    for _ in range(next_words):

        # Convert text → sequence
        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        # Pad sequence
        token_list = pad_sequences([token_list],
                                   maxlen=max_lenght-1,
                                   padding='pre')

        # Predict probabilities
        predicted_probs = model.predict(token_list, verbose=0)[0]

        # 🔥 Use Top-K + Temperature Sampling
        predicted_index = top_k_sampling(predicted_probs, k)

        # Convert index → word
        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted_index:
                output_word = word
                break

        # 🔥 Skip unwanted frequent words
        if output_word in ignore_words:
            continue

        # Append predicted word
        seed_text += " " + output_word

    return seed_text
```
